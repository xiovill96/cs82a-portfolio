```python
def multiply_with_while_loop(number, iterations):
    result = 0
    count = 0
    while count < iterations:
        result += number
        count += 1
    return result

num1 = 5
num2 = 8
product = multiply_with_while_loop(num1, num2)
print(f"The result of {num1} and {num2} is: {product}")
```

```python
def multiply_with_loop(number, iterations):
    result = 0
    for _ in range(iterations):
        result += number
    return result
num1 = 7
num2 = 9
product = multiply_with_while_loop(num1, num2)
print(f"The result of {num1} and {num2} is: {product}")
```
