```python
temperatures = [10, 16, 23, 28, 14, 22, 30]

for temp in temperatures:
    
    if temp <= 15:
        label = "Cold"
    elif 15 < temp <= 25:
        label = "Warm"
    else:
        label = "Hot"
    
    
    print(f"{temp}°C: {label}")
```

I wanted to only do 3 cutoffs so I chose anything under 15 to be "cold". Anything between 15 and 25 as warm. With anything above 25 to be considered hot.
